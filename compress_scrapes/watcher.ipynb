```python
import sys
sys.path.append("../") # go to parent dir

from custom_helpers_py.utilities import camel_to_snake_case
import pandas as pd
from os import listdir
from os.path import join
import json
```

```python
COMPRESSED_FOLDER_PATH = join("../.outFiles/analysis/compressed")
SENATE_FILE_PATH = join("../.outFiles/watcher/senate_watcher.csv")
HOUSE_FILE_PATH = join("../.outFiles/watcher/house_watcher.csv")

def get_doc_id(in_str:str):
    in_str = in_str.removesuffix("/")
    last_slash_idx = in_str.rfind("/")
    doc_id = in_str[last_slash_idx + 1:]
    doc_id = doc_id.removesuffix(".pdf")
    return doc_id
    
senate_df = pd.read_csv(SENATE_FILE_PATH, encoding="utf-8")
senate_df["doc_id"] = senate_df["ptr_link"].apply(get_doc_id)
senate_df = senate_df.drop(columns=["ptr_link"])
senate_df.columns = [
    "tx_date",
    "owner",
    "ticker",
    "asset_name",
    "asset_type",
    "tx_type",
    "amount",
    "comments",
    "p_party",
    "p_state",
    "asset_industry",
    "asset_sector",
    "p_full_name",
    "notif_date",
    "doc_id"
]
senate_df["data_source"] = "WATCHER_SENATE"

to_save_path = join(COMPRESSED_FOLDER_PATH, "watcher_senate.csv")
senate_df.to_csv(to_save_path, index=False, header=True)

senate_df
```

```python
house_df = pd.read_csv(HOUSE_FILE_PATH, encoding="utf-8")

house_df["doc_id"] = house_df["ptr_link"].apply(get_doc_id)
house_df = house_df.drop(columns=["disclosure_year", "ptr_link"])
house_df.columns = [
    "notif_date",
    "tx_date",
    "owner",
    "ticker",
    "asset_name",
    "tx_type",
    "amount",
    "p_full_name",
    "p_district",
    "p_state",
    "cap_gains_over_200_usd",
    "asset_industry",
    "asset_sector",
    "p_party",
    "doc_id"
]
house_df["data_source"] = "WATCHER_HOUSE"

to_save_path = join(COMPRESSED_FOLDER_PATH, "watcher_house.csv")
house_df.to_csv(to_save_path, index=False, header=True)

house_df
```
