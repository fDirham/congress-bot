```python
import pandas as pd
from os import listdir
from os.path import join
import json
```

```python
COMPRESSED_FOLDER_PATH = join("../.outFiles/analysis/compressed")

LIST_FP = join("../.outFiles/congress_members/list.json")
politicians_df = pd.read_json(LIST_FP)
politicians_df
```

```python
politicians_df = politicians_df.drop(columns=["heading", "served"])
politicians_df.columns = [
    "p_full_name",
    "p_chamber",
    "p_state",
    "p_district",
    "p_party",
]

politicians_df["p_party"] = politicians_df["p_party"].str.slice(0,1)
politicians_df["p_chamber"] = politicians_df["p_chamber"].str.slice(0,1)

def fix_name(in_name: str):
    if "," in in_name:
        tmp = in_name.split(",")
        in_name = tmp[1] + " " + tmp[0]
    
    in_name = in_name.replace(".","")
    return in_name

politicians_df["p_full_name"] = politicians_df["p_full_name"].apply(fix_name) 

save_path = join(COMPRESSED_FOLDER_PATH, "politician.csv")
politicians_df.to_csv(save_path, index=False, header=True)


politicians_df
```
