```python
import sys
sys.path.append("../") # go to parent dir

from custom_helpers_py.utilities import camel_to_snake_case
import pandas as pd
from os import listdir
from os.path import join
import json
```

```python
COMPRESSED_FOLDER_PATH = join("../.outFiles/analysis/compressed")

IN_FOLDER_PATH = join("../.outFiles/trendspider/scrapes")
file_list = listdir(IN_FOLDER_PATH)
df_list = []
for file_name in file_list:
    if not file_name.endswith(".json"):
        continue
    with open(join(IN_FOLDER_PATH, file_name), "r", encoding="utf-8") as in_file:
        tmp = json.loads(in_file.read())
    
    data_list = tmp["dataList"]
    df = pd.DataFrame(data_list)
    
    df_list.append(df)

master_df = pd.concat(df_list)
master_df = master_df.drop(columns=["excessReturn"])
master_df.columns = [camel_to_snake_case(col) for col in master_df.columns]
master_df = master_df.rename(columns={
    "traded": "tx_date",
    "filed": "notif_date",
    "politician_name": "p_full_name",
    "party": "p_party",
    "chamber": "p_chamber",
})

to_save_path = join(COMPRESSED_FOLDER_PATH, "trendspider.csv")
master_df.to_csv(to_save_path, index=False, header=True)

master_df
```
